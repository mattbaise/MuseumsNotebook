```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("museums.csv", low_memory=False)
df.head()
df.columns
df.shape
```

```python
df["State (Administrative Location)"].value_counts()
```

```python
df["Museum Type"].unique()
```

```python
zoos_aquariums = df[
    df["Museum Type"] == "ZOO, AQUARIUM, OR WILDLIFE CONSERVATION"
    ]
len(zoos_aquariums)
```

```python
zoos = df[
    df["Museum Name"].str.contains("Zoo", case=False, na=False)
    ]
len(zoos)
```

```python
aquariums = df[
    df["Museum Name"].str.contains("Aquarium", case=False, na=False)
    ]
len(aquariums)
```

```python
df.sort_values(
    by="Revenue",
    ascending=False
)[["Museum Name","Revenue"]].head(1)
```

```python
museum_types = df.groupby([
    "State (Administrative Location)",
    "Museum Type"
]).size()

museum_types
```
